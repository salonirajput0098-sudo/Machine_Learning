```python
import numpy as np
from sklearn.preprocessing import PowerTransformer
```

```python
# A right-skewed features

X = np.array([[-1],[0],[5],[5],[10],[900]])

# Create the transformer

pt = PowerTransformer(method="yeo-johnson")

# fit and transformer

X_tansformed = pt.fit_transform(X)

print("Original:")
print(X.ravel())

print("\nTransformed:")
print(np.round(X_tansformed.ravel(),3))

print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
# A right-skewed features

y = np.array([[1],[4],[3],[5],[10],[50]])

# Create the transformer

pt = PowerTransformer(method="box-cox")

# fit and transformer

y_tansformed = pt.fit_transform(y)

print("Original:")
print(y.ravel())

print("\nTransformed:")
print(np.round(y_tansformed.ravel(),3))

print("\nLearned lambda:")
print(pt.lambdas_)
```
