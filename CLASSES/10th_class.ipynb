```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.3 , random_state = 23 , shuffle= False)
```

```python
from sklearn.metrics import (accuracy_score , precision_score,
recall_score , classification_report , confusion_matrix,)
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
for i in range(1,6):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(x_train , y_train)
    y_pred = knn.predict(x_test)
    print(f'i----------------{i}')
    print(confusion_matrix(y_test,y_pred))
    print('-----------------')
    print(accuracy_score(y_test,y_pred))
    print('------------------')
    print(precision_score(y_test,y_pred , average='weighted'))
    print('-----------------')
    print(classification_report(y_test,y_pred))
```

```python
knn = KNeighborsClassifier()
param = {'n_neighbors':[2,3,4,5,6,7],
'p':[1,2],

'algorithm':['auto','ball_tree','brute','kd_tree']}
```

```python
GridSearchCV(knn,param_grid=param,cv=5)
```

```python
GridSearchCV(knn,param_grid=param,cv=5).fit(x_train,y_train).best_params_
```

```python
knn = KNeighborsClassifier(n_neighbors=3 , p=1 , algorithm='auto')
knn.fit(x_train , y_train)
y_pred = knn.predict(x_test)
print(confusion_matrix(y_test,y_pred))
print('-----------------')
print(accuracy_score(y_test,y_pred))
print('------------------')
print(precision_score(y_test,y_pred , average='weighted'))
print('-----------------')
print(classification_report(y_test,y_pred))
```
