```python
array = np.linspace(12,22,num=15,dtype=int)
```

```python
array
```

```python
def variance_std (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance = total/(len(args))
    return calculated_variance , calculated_variance**(1/2)
```

```python
variance_std(*array)
```

```python
np.var(array)
```

```python
def sample_variance_std (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance = total/(len(args)-1)
    return calculated_variance , calculated_variance**(1/2)
```

```python
sample_variance_std(*array)
```

```python
import matplotlib.pyplot as plt
```

```python
hours = [1,2,3,4,5,6]
marks = [30,40,45,55,65,75]
```

```python
plt.figure(figsize=(3,3))
plt.scatter(hours,marks,marker='+',color='red')
plt.xlabel('no of hours')
plt.ylabel('marks')
plt.title('Hours vs Marks')
plt.show()
```

```python
products = ["laptop","mobile","tablet","desktop"]
prices = [40, 80, 30, 60]
```

```python
plt.figure(figsize=(3,3))
plt.bar(products,prices,color='green')
plt.xlabel('Products')
plt.ylabel('Prices')
plt.title('Product Prices')
plt.tight_layout()
plt.show()
```
