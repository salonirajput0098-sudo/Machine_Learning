Question 1: OLS Mathematical Challenge

```python
import numpy as np

x = np.array([1, 2, 4, 5, 7, 8], dtype=float)
y = np.array([3, 4, 7, 8, 12, 10], dtype=float)
```

```python
import numpy as np

x = np.array([1, 2, 4, 5, 7, 8], dtype=float)
y = np.array([3, 4, 7, 8, 12, 10], dtype=float)

# Calculate means
x_mean = np.mean(x)
y_mean = np.mean(y)

# Calculate slope
m = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)

# Calculate intercept
b = y_mean - m * x_mean

# Regression equation
print("Mean of x:", x_mean)
print("Mean of y:", y_mean)
print("Slope:", m)
print("Intercept:", b)
print("Equation: y =", m, "x +", b)
```

```python
# Calculate predictions
y_pred = b + m * x

# Calculate residuals
residuals = y - y_pred

# Calculate SSE
SSE = np.sum(residuals ** 2)

# Calculate MSE
MSE = SSE / len(y)

print("Predicted values:", y_pred)
print("Residuals:", residuals)
print("SSE:", SSE)
print("MSE:", MSE)
```

Question 2: Build a Regression Model from Scratch

```python
import numpy as np

x = np.array([2, 4, 6, 8, 10, 12], dtype=float)
y = np.array([15, 20, 28, 35, 43, 48], dtype=float)
```

```python
import numpy as np

x = np.array([2, 4, 6, 8, 10, 12], dtype=float)
y = np.array([15, 20, 28, 35, 43, 48], dtype=float)

# Calculate mean
x_mean = np.mean(x)
y_mean = np.mean(y)

# Calculate slope
m = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)

# Calculate intercept
b = y_mean - m * x_mean

print("Slope:", m)
print("Intercept:", b)

# Regression equation
print("Regression equation: y =", m, "x +", b)
```

```python
# New input values
new_x = np.array([5, 9, 16], dtype=float)

# Prediction
predictions = b + m * new_x

print("Predictions:", predictions)
```

```python
# Predict on original dataset
y_pred = b + m * x

# Residuals
residuals = y - y_pred

# SSE
SSE = np.sum(residuals ** 2)

# MSE
MSE = SSE / len(y)

# RMSE
RMSE = np.sqrt(MSE)

print("Predicted values:", y_pred)
print("Residuals:", residuals)
print("SSE:", SSE)
print("MSE:", MSE)
print("RMSE:", RMSE)
```

Question 3: Regression Model Selection and Extrapolation Risk

```python
import numpy as np

hours = np.array([40, 85, 120, 200], dtype=float)

fuel = 25 + 1.8 * hours

for h, f in zip(hours, fuel):
    if 10 <= h <= 100:
        prediction_type = "Interpolation"
    else:
        prediction_type = "Extrapolation"

    print(h, "hours:", f, "-", prediction_type)
```
