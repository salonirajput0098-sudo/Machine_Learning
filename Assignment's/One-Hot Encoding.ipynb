Q1. Customer Data Encoding and Feature Analysis

An e-commerce company wants to prepare customer data for a machine learning model.



```python

import pandas as pd

df = pd.DataFrame({
    "Customer_ID": range(101, 113),
    "City": [
        "Bhopal", "Delhi", "Mumbai", "Bhopal",
        "Delhi", "Pune", "Mumbai", "Pune",
        "Bhopal", "Delhi", "Pune", "Mumbai"
    ],
    "Payment_Method": [
        "UPI", "Cash", "Card", "UPI",
        "Card", "Cash", "UPI", "Card",
        "Cash", "UPI", "Card", "Cash"
    ],
    "Purchase_Amount": [
        500, 1200, 800, 1500, 700, 900,
        1800, 1100, 600, 1300, 950, 2000
    ]
})

# Apply One-Hot Encoding
encoded_df = pd.get_dummies(
    df,
    columns=["City", "Payment_Method"],
    dtype=int
)

print("Encoded DataFrame:")
print(encoded_df)

# Verify number of rows
print("\nOriginal Rows:", len(df))
print("Encoded Rows:", len(encoded_df))

# Calculate new columns
new_columns = len(encoded_df.columns) - len(df.columns)

print("New Columns Created:", new_columns)
```

```python

# Original DataFrame dimensions
print("Original Shape:", df.shape)

# Encoded DataFrame dimensions
print("Encoded Shape:", encoded_df.shape)

# Memory usage
original_memory = df.memory_usage(deep=True).sum()

encoded_memory = encoded_df.memory_usage(deep=True).sum()

print("\nOriginal Memory Usage:",
      original_memory, "bytes")

print("Encoded Memory Usage:",
      encoded_memory, "bytes")

# Compare memory
if encoded_memory > original_memory:
    print("\nEncoded DataFrame uses more memory.")
else:
    print("\nEncoded DataFrame uses less memory.")
```

Q2. One-Hot Encoding in a Machine Learning Pipeline

```python

import pandas as pd

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

df = pd.DataFrame({
    "Age": [22, 35, 28, 45, 31, 24, 40, 29, 38, 26, 33, 42],
    "City": [
        "Bhopal", "Delhi", "Mumbai", "Pune",
        "Delhi", "Bhopal", "Mumbai", "Pune",
        "Delhi", "Mumbai", "Bhopal", "Pune"
    ],
    "Browser": [
        "Chrome", "Firefox", "Safari", "Chrome",
        "Safari", "Firefox", "Chrome", "Safari",
        "Firefox", "Chrome", "Safari", "Firefox"
    ],
    "Purchase_Amount": [
        600, 1500, 900, 2200, 1300, 750,
        2000, 1100, 1800, 850, 1400, 2100
    ]
})

# Separate features
X = df[["Age", "City", "Browser"]]

# Create ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical",
         OneHotEncoder(handle_unknown="ignore"),
         ["City", "Browser"])
    ],
    remainder="passthrough"
)

# Transform data
X_encoded = preprocessor.fit_transform(X)

# Get feature names
feature_names = preprocessor.get_feature_names_out()

print("Feature Names:")
print(feature_names)

print("\nTransformed Matrix:")
print(X_encoded)

print("\nShape:", X_encoded.shape)

# Verify numerical data
print("\nAll values are numerical:",
      not pd.isna(X_encoded).any())
```

```python

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# Features and target
X = df[["Age", "City", "Browser"]]
y = df["Purchase_Amount"]

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42
)

# Create pipeline
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

# Train model
model.fit(X_train, y_train)

# Predict values
y_pred = model.predict(X_test)

print("Actual Values:")
print(y_test.values)

print("\nPredicted Values:")
print(y_pred)

# Calculate MAE
mae = mean_absolute_error(y_test, y_pred)

print("\nMean Absolute Error:", mae)
```

Q3. High Cardinality and Memory Consumption (10 Marks)

```python

import pandas as pd

df = pd.DataFrame({
    "Product_Category": [
        "Electronics", "Clothing", "Books", "Electronics",
        "Food", "Books", "Clothing", "Food"
    ],
    "City": [
        "Bhopal", "Delhi", "Mumbai", "Pune",
        "Indore", "Delhi", "Bhopal", "Jaipur"
    ],
    "User_ID": [
        "U101", "U102", "U103", "U104",
        "U105", "U106", "U107", "U108"
    ],
    "Purchase_Amount": [
        1200, 800, 400, 2500, 600, 350, 900, 700
    ]
})

# Check unique values
print("Unique Values:")

for col in ["Product_Category", "City", "User_ID"]:
    print(col, ":", df[col].nunique())

# Apply One-Hot Encoding
encoded_df = pd.get_dummies(
    df,
    columns=["Product_Category", "City", "User_ID"],
    dtype=int
)

print("\nEncoded DataFrame:")
print(encoded_df)

# Compare dimensions
print("\nOriginal Shape:", df.shape)
print("Encoded Shape:", encoded_df.shape)
```

```python

import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Original memory usage
original_memory = df.memory_usage(deep=True).sum()

# Encoded memory usage
encoded_memory = encoded_df.memory_usage(deep=True).sum()

print("Original Memory:",
      original_memory, "bytes")

print("Encoded Memory:",
      encoded_memory, "bytes")

# Remove User_ID because it is an identifier
df_clean = df.drop(columns=["User_ID"])

print("\nDataFrame after removing User_ID:")
print(df_clean)

# Apply OneHotEncoder with sparse output
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=True
)

# Encode categorical columns
encoded_data = encoder.fit_transform(
    df_clean[["Product_Category", "City"]]
)

print("\nSparse Matrix:")
print(encoded_data)

print("\nSparse Matrix Shape:",
      encoded_data.shape)

print("Number of non-zero values:",
      encoded_data.nnz)

print("Memory Used by Sparse Matrix:",
      encoded_data.data.nbytes, "bytes")
```
