Question 1: Multiple Linear Regression and Coefficient Interpretation

```python
import pandas as pd

# Create dataset
df = pd.DataFrame({
    'House': ['A', 'B', 'C', 'D'],
    'Area': [1200, 1800, 2400, 1500],
    'Bedrooms': [2, 3, 4, 3],
    'Age': [10, 5, 8, 12],
    'Location_Score': [3, 4, 5, 2]
})

# Calculate predicted price
df['Predicted_Price'] = (
    12
    + 0.045 * df['Area']
    + 6 * df['Bedrooms']
    - 1.2 * df['Age']
    + 8 * df['Location_Score']
)

print(df)
```
