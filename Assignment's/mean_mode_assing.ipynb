Q2. Customer Spending Analysis Using the Tips Dataset
(a) Calculate Mean, Median and all Modes of total_bill and tip. Compare Lunch and Dinner groups.

```python

import seaborn as sns
import pandas as pd

df = sns.load_dataset('tips')

def calculate_modes(data):
    frequency = {}

    for value in data:
        frequency[value] = frequency.get(value, 0) + 1

    max_frequency = max(frequency.values())

    if max_frequency == 1:
        return []

    return [k for k, v in frequency.items()
            if v == max_frequency]

def mean(data):
    return sum(data) / len(data)

def median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 0:
        return (data[n//2-1] + data[n//2]) / 2
    return data[n//2]

print("TOTAL BILL")
print("Mean:", mean(df['total_bill'].tolist()))
print("Median:", median(df['total_bill'].tolist()))
print("Mode:", calculate_modes(df['total_bill'].tolist()))

print("\nTIP")
print("Mean:", mean(df['tip'].tolist()))
print("Median:", median(df['tip'].tolist()))
print("Mode:", calculate_modes(df['tip'].tolist()))

for group in ['Lunch', 'Dinner']:

    group_data = df[df['time'] == group]['total_bill'].tolist()

    print("\n", group)
    print("Mean:", mean(group_data))
    print("Median:", median(group_data))
```

Q3. Build a Central Tendency Calculator From Scratch

(a) Build reusable Python functions for Mean, Median and Mode.

```python

# Function to calculate Mean
def calculate_mean(data):
    return sum(data) / len(data)


# Function to calculate Median
def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 0:
        return (data[n//2 - 1] + data[n//2]) / 2
    else:
        return data[n//2]


# Function to calculate Mode
def calculate_mode(data):
    frequency = {}

    for num in data:
        frequency[num] = frequency.get(num, 0) + 1

    max_frequency = max(frequency.values())

    if max_frequency == 1:
        return "No Mode"

    return [num for num, count in frequency.items()
            if count == max_frequency]


# Five datasets
data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20]
data_3 = [7, 14, 21, 28, 35]
data_4 = [100, 100, 100, 100]
data_5 = [1, 2, 3, 4, 5, 100]

datasets = [data_1, data_2, data_3, data_4, data_5]

# Generate report
for i, data in enumerate(datasets, start=1):

    print("\nDataset", i)
    print("Data:", data)
    print("Mean:", calculate_mean(data))
    print("Median:", calculate_median(data))
    print("Mode:", calculate_mode(data))
```

Q4. Statistical Transformations and Reasoning

(a) Calculate Mean, Median and Mode. Remove the largest observation and recalculate

```python

orders = [
    1, 1, 1, 2, 2, 2, 2, 3, 3, 4,
    4, 5, 5, 6, 8, 10, 15, 20, 25, 80
]

# Central tendency functions
def mean(data):
    return sum(data) / len(data)

def median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 0:
        return (data[n//2-1] + data[n//2]) / 2
    return data[n//2]

def mode(data):
    freq = {}

    for x in data:
        freq[x] = freq.get(x, 0) + 1

    maximum = max(freq.values())

    if maximum == 1:
        return "No Mode"

    return [x for x, count in freq.items()
            if count == maximum]


# Original dataset
print("Original Data")
print("Mean:", mean(orders))
print("Median:", median(orders))
print("Mode:", mode(orders))

# Remove largest observation
new_orders = orders.copy()
new_orders.remove(max(new_orders))

print("\nAfter Removing Largest Value")
print("Mean:", mean(new_orders))
print("Median:", median(new_orders))
print("Mode:", mode(new_orders))
```

(b) Create two transformed datasets.

```python

# Original measures
original_mean = mean(orders)
original_median = median(orders)
original_mode = mode(orders)

# Transformation 1: Add 20
data_add = [x + 20 for x in orders]

# Transformation 2: Multiply by 3
data_multiply = [x * 3 for x in orders]


# Display original measures
print("Original Dataset")
print("Mean:", original_mean)
print("Median:", original_median)
print("Mode:", original_mode)


# Display addition transformation
print("\nAfter Adding 20")
print("Mean:", mean(data_add))
print("Median:", median(data_add))
print("Mode:", mode(data_add))


# Display multiplication transformation
print("\nAfter Multiplying by 3")
print("Mean:", mean(data_multiply))
print("Median:", median(data_multiply))
print("Mode:", mode(data_multiply))


# Verify transformation rules
print("\nVerification")

print("Mean + 20:", original_mean + 20)
print("Median + 20:", original_median + 20)
print("Mode + 20:", [x + 20 for x in original_mode])

print("Mean * 3:", original_mean * 3)
print("Median * 3:", original_median * 3)
print("Mode * 3:", [x * 3 for x in original_mode])
```

Q5. Missing Values and Outlier Investigation

(a) Convert data into Pandas Series, identify missing values and calculate Mean, Median, Mode and IQR.

```python

import pandas as pd

sales = [
    12, 15, 14, 16, None,
    15, 18, 14, 17, 16,
    None, 19, 15, 120, 14
]

# Convert into Pandas Series
s = pd.Series(sales)

# Identify missing values
print("Missing Values:")
print(s.isnull())

missing_count = s.isnull().sum()
total_count = len(s)

percentage = (missing_count / total_count) * 100

print("\nMissing Count:", missing_count)
print("Missing Percentage:", percentage)

# Remove missing values
clean_sales = s.dropna()

# Manual Mean
def calculate_mean(data):
    return sum(data) / len(data)

# Manual Median
def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 0:
        return (data[n//2-1] + data[n//2]) / 2
    return data[n//2]

# Manual Mode
def calculate_mode(data):
    freq = {}

    for x in data:
        freq[x] = freq.get(x, 0) + 1

    maximum = max(freq.values())

    if maximum == 1:
        return "No Mode"

    return [x for x, count in freq.items()
            if count == maximum]


# Central tendency
print("\nMean:", calculate_mean(clean_sales.tolist()))
print("Median:", calculate_median(clean_sales.tolist()))
print("Mode:", calculate_mode(clean_sales.tolist()))

# IQR Method
Q1 = clean_sales.quantile(0.25)
Q3 = clean_sales.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Identify outliers
outliers = clean_sales[
    (clean_sales < lower_limit) |
    (clean_sales > upper_limit)
]

print("Outliers:", outliers.tolist())
```

(b) Remove suspected outlier 120 and recalculate all measures.

```python

# Original measures
old_mean = calculate_mean(clean_sales.tolist())
old_median = calculate_median(clean_sales.tolist())

# Remove suspected outlier
new_sales = clean_sales[clean_sales != 120]

# Recalculate measures
new_mean = calculate_mean(new_sales.tolist())
new_median = calculate_median(new_sales.tolist())
new_mode = calculate_mode(new_sales.tolist())

print("After Removing Outlier")

print("Mean:", new_mean)
print("Median:", new_median)
print("Mode:", new_mode)

# Percentage change
mean_change = ((new_mean - old_mean) / old_mean) * 100

median_change = ((new_median - old_median) / old_median) * 100

print("\nMean Percentage Change:", mean_change)
print("Median Percentage Change:", median_change)

# Comparison
print("\nBefore Removing Outlier")
print("Mean:", old_mean)
print("Median:", old_median)

print("\nAfter Removing Outlier")
print("Mean:", new_mean)
print("Median:", new_median)
```
