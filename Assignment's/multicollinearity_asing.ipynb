Create a dataset containing at least four numerical features related to house prices.



```python
import pandas as pd
import numpy as np
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
```

```python
df = pd.DataFrame({
    "Area_sqft": [1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800, 3000, 3200, 3500, 3800],
    "Area_sqm": [93.0, 111.5, 139.4, 167.4, 185.8, 204.4, 232.5, 260.3, 278.7, 297.3, 325.5, 353.4],
    "Bedrooms": [2, 2, 3, 3, 3, 4, 4, 4, 5, 5, 5, 6],
    "Bathrooms": [1, 2, 2, 2, 3, 3, 3, 4, 4, 4, 5, 5],
    "Floors": [1, 1, 1, 2, 2, 2, 2, 2, 3, 3, 3, 3],
    "Age": [20, 18, 15, 12, 10, 8, 7, 6, 5, 4, 3, 2],
    "Price": [30, 35, 45, 55, 60, 68, 75, 85, 95, 105, 120, 135]
})

print(df)
```

```python
corr_matrix = df.corr()

print(corr_matrix.round(2))
```

```python
corr_matrix = df.drop("Price", axis=1).corr()

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > 0.8:
            print(
                corr_matrix.columns[i],
                "and",
                corr_matrix.columns[j],
                "=", round(corr_matrix.iloc[i, j], 2)
            )
```

```python
X = df.drop("Price", axis=1)

X = add_constant(X)

vif_data = pd.DataFrame()
vif_data["Feature"] = X.columns
vif_data["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

vif_data = vif_data[vif_data["Feature"] != "const"]

print(vif_data)
```

```python
# Area in sqft and sqm represent the same information
# Therefore, Area_sqm is redundant

df = df.drop("Area_sqm", axis=1)

print(df.head())
print("Remaining columns:", df.columns)
```
