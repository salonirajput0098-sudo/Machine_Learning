Q1. Custom Feature Transformation Using FunctionTransformer

```python

import numpy as np
import pandas as pd

from sklearn.preprocessing import FunctionTransformer

df = pd.DataFrame({
    "Customer_ID": range(1, 11),
    "Income": [15000, 18000, 22000, 25000, 30000,
               35000, 42000, 55000, 120000, 500000],
    "Monthly_Expenses": [8000, 9000, 11000, 12000, 15000,
                         17000, 20000, 25000, 45000, 180000]
})

# Select numerical features
X = df[["Income", "Monthly_Expenses"]]

# Create FunctionTransformer
transformer = FunctionTransformer(np.log1p)

# Apply transformation
X_transformed = transformer.fit_transform(X)

transformed_df = pd.DataFrame(
    X_transformed,
    columns=["Income_Log", "Expenses_Log"]
)

print("Original Data:")
print(X)

print("\nTransformed Data:")
print(transformed_df)

# Compare statistics
print("\nOriginal Mean:")
print(X.mean())

print("\nTransformed Mean:")
print(transformed_df.mean())

print("\nOriginal Range:")
print(X.max() - X.min())

print("\nTransformed Range:")
print(transformed_df.max() - transformed_df.min())

print("\nOriginal Standard Deviation:")
print(X.std())

print("\nTransformed Standard Deviation:")
print(transformed_df.std())
```

```python

import pandas as pd
import numpy as np

from sklearn.preprocessing import FunctionTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error

df = pd.DataFrame({
    "Income": [15000, 18000, 22000, 25000, 30000,
               35000, 42000, 55000, 120000, 500000],
    "Monthly_Expenses": [8000, 9000, 11000, 12000, 15000,
                         17000, 20000, 25000, 45000, 180000]
})

X = df[["Income"]]
y = df["Monthly_Expenses"]

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Model without transformation
model1 = LinearRegression()
model1.fit(X_train, y_train)

pred1 = model1.predict(X_test)

# Model with log transformation
model2 = Pipeline([
    ("log", FunctionTransformer(np.log1p)),
    ("regression", LinearRegression())
])

model2.fit(X_train, y_train)

pred2 = model2.predict(X_test)

# Compare performance
print("Original Model MAE:",
      mean_absolute_error(y_test, pred1))

print("Log Model MAE:",
      mean_absolute_error(y_test, pred2))

print("\nOriginal Model RMSE:",
      np.sqrt(mean_squared_error(y_test, pred1)))

print("Log Model RMSE:",
      np.sqrt(mean_squared_error(y_test, pred2)))
```

Q2. Designing and Validating Custom Transformations

```python

import pandas as pd
import numpy as np

from sklearn.preprocessing import FunctionTransformer

df = pd.DataFrame({
    "Sales": [100, 150, 200, 250, 400, 600, 900, 1500, 3000, 8000],
    "Transaction_Count": [1, 2, 2, 3, 4, 5, 7, 10, 15, 30],
    "Discount_Percent": [0, 5, 10, 15, 20, 5, 10, 25, 30, 40],
    "Revenue": [120, 170, 230, 280, 450, 650, 950, 1600, 3200, 8500]
})

# Function to convert sales into thousands
def sales_to_thousands(X):
    return X / 1000

# Create transformers
sales_transformer = FunctionTransformer(
    sales_to_thousands
)

sqrt_transformer = FunctionTransformer(
    np.sqrt
)

# Apply transformations
df["Sales_Thousands"] = sales_transformer.transform(
    df[["Sales"]]
)

df["Transaction_Sqrt"] = sqrt_transformer.transform(
    df[["Transaction_Count"]]
)

print(df)
```

```python

import pandas as pd
import numpy as np

from sklearn.preprocessing import FunctionTransformer

df = pd.DataFrame({
    "Sales": [100, 150, 200, 250, 400, 600, 900, 1500, 3000, 8000]
})

# Create reusable clipping function
def clip_values(X, lower, upper):
    return np.clip(X, lower, upper)

# First clipping range
clip1 = FunctionTransformer(
    clip_values,
    kw_args={"lower": 100, "upper": 3000}
)

# Second clipping range
clip2 = FunctionTransformer(
    clip_values,
    kw_args={"lower": 200, "upper": 1500}
)

# Apply clipping
df["Clip_1"] = clip1.transform(df[["Sales"]])
df["Clip_2"] = clip2.transform(df[["Sales"]])

print(df)

# Compare statistics
print("\nOriginal Statistics:")
print(df["Sales"].agg(["mean", "std", "min", "max"]))

print("\nClipping Range 100-3000:")
print(df["Clip_1"].agg(["mean", "std", "min", "max"]))

print("\nClipping Range 200-1500:")
print(df["Clip_2"].agg(["mean", "std", "min", "max"]))
```

Q3. PowerTransformer and Distribution Analysis 

```python

import pandas as pd
from sklearn.preprocessing import PowerTransformer

df = pd.DataFrame({
    "Annual_Income": [20000, 22000, 25000, 28000, 32000,
                      40000, 50000, 70000, 150000, 600000],

    "Loan_Amount": [5000, 7000, 9000, 12000, 15000,
                    20000, 30000, 50000, 100000, 400000],

    "Account_Balance": [-500, 0, 200, 500, 1000,
                        2500, 4000, 8000, 15000, 50000]
})

# Create PowerTransformer
pt = PowerTransformer(method="yeo-johnson")

# Apply transformation
transformed_data = pt.fit_transform(df)

transformed_df = pd.DataFrame(
    transformed_data,
    columns=df.columns
)

print("Transformed Data:")
print(transformed_df)

# Compare skewness
print("\nOriginal Skewness:")
print(df.skew())

print("\nTransformed Skewness:")
print(transformed_df.skew())

# Display lambda values
print("\nLearned Lambda Values:")
print(pd.Series(pt.lambdas_, index=df.columns))
```

```python

import pandas as pd
from sklearn.preprocessing import PowerTransformer

df = pd.DataFrame({
    "Annual_Income": [20000, 22000, 25000, 28000, 32000,
                      40000, 50000, 70000, 150000, 600000],

    "Loan_Amount": [5000, 7000, 9000, 12000, 15000,
                    20000, 30000, 50000, 100000, 400000],

    "Account_Balance": [-500, 0, 200, 500, 1000,
                        2500, 4000, 8000, 15000, 50000]
})

# Apply Box-Cox to positive columns
boxcox = PowerTransformer(method="box-cox")

positive_columns = ["Annual_Income", "Loan_Amount"]

boxcox_data = boxcox.fit_transform(
    df[positive_columns]
)

boxcox_df = pd.DataFrame(
    boxcox_data,
    columns=positive_columns
)

print("Box-Cox Transformed Data:")
print(boxcox_df)

print("\nBox-Cox Lambda Values:")
print(pd.Series(
    boxcox.lambdas_,
    index=positive_columns
))

# Attempt Box-Cox on Account_Balance
try:
    boxcox.fit_transform(df[["Account_Balance"]])

except ValueError as e:
    print("\nBox-Cox Error:")
    print(e)

# Safe preprocessing
result = df.copy()

for col in df.columns:

    if (df[col] > 0).all():

        transformer = PowerTransformer(
            method="box-cox"
        )

        result[col] = transformer.fit_transform(
            df[[col]]
        ).ravel()

    else:

        transformer = PowerTransformer(
            method="yeo-johnson"
        )

        result[col] = transformer.fit_transform(
            df[[col]]
        ).ravel()

print("\nSafely Transformed Data:")
print(result)
```
