Q1. Employee Performance: Population vs. Sample Variance

```python
scores = [45, 50, 52, 55, 58, 60, 65, 75]
```

```python

scores = [45, 50, 52, 55, 58, 60, 65, 75]

# Calculate Mean
def calculate_mean(data):
    return sum(data) / len(data)

# Calculate Population Variance
def population_variance(data):
    mean = calculate_mean(data)
    squared_diff = [(x - mean) ** 2 for x in data]

    return sum(squared_diff) / len(data)

# Calculate Sample Variance
def sample_variance(data):
    mean = calculate_mean(data)
    squared_diff = [(x - mean) ** 2 for x in data]

    return sum(squared_diff) / (len(data) - 1)

# Calculate Standard Deviations
def population_std(data):
    return population_variance(data) ** 0.5

def sample_std(data):
    return sample_variance(data) ** 0.5


print("Mean:", calculate_mean(scores))
print("Population Variance:", population_variance(scores))
print("Sample Variance:", sample_variance(scores))
print("Population Standard Deviation:", population_std(scores))
print("Sample Standard Deviation:", sample_std(scores))
```


Q2. Customer Spending: Comparing Data Spread

```python

import seaborn as sns

# Load dataset
df = sns.load_dataset('tips')

# Function to calculate mean
def calculate_mean(data):
    return sum(data) / len(data)

# Population variance
def population_variance(data):
    mean = calculate_mean(data)

    return sum((x - mean) ** 2 for x in data) / len(data)

# Population standard deviation
def population_std(data):
    return population_variance(data) ** 0.5

# Coefficient of Variation
def calculate_cv(data):
    return (population_std(data) / calculate_mean(data)) * 100


# Select columns
bill = df['total_bill'].tolist()
tip = df['tip'].tolist()

# Total Bill Analysis
print("TOTAL BILL")
print("Mean:", calculate_mean(bill))
print("Variance:", population_variance(bill))
print("Standard Deviation:", population_std(bill))
print("CV:", calculate_cv(bill))

# Tip Analysis
print("\nTIP")
print("Mean:", calculate_mean(tip))
print("Variance:", population_variance(tip))
print("Standard Deviation:", population_std(tip))
print("CV:", calculate_cv(tip))
```

(b) Identify the five largest bills and temporarily exclude them.
Python Code

```python

# Original statistics
original_mean = calculate_mean(bill)
original_variance = population_variance(bill)
original_std = population_std(bill)

# Find five largest bills
largest_bills = df.nlargest(5, 'total_bill')

print("Five Largest Bills:")
print(largest_bills['total_bill'].tolist())

# Remove these observations temporarily
new_df = df.drop(largest_bills.index)

new_bill = new_df['total_bill'].tolist()

# Calculate new statistics
new_mean = calculate_mean(new_bill)
new_variance = population_variance(new_bill)
new_std = population_std(new_bill)

print("\nBefore Exclusion")
print("Mean:", original_mean)
print("Variance:", original_variance)
print("Standard Deviation:", original_std)

print("\nAfter Exclusion")
print("Mean:", new_mean)
print("Variance:", new_variance)
print("Standard Deviation:", new_std)

# Calculate percentage changes
print("\nPercentage Changes")

print("Mean:", ((new_mean-original_mean)/original_mean)*100)

print("Variance:", ((new_variance-original_variance)/original_variance)*100)

print("Standard Deviation:", ((new_std-original_std)/original_std)*100)
```

(a) Implement separate functions for all calculations.

```python

# Function to calculate Mean
def calculate_mean(data):
    return sum(data) / len(data)


# Population Variance
def population_variance(data):
    mean = calculate_mean(data)

    total = 0
    for x in data:
        total += (x - mean) ** 2

    return total / len(data)


# Sample Variance
def sample_variance(data):
    mean = calculate_mean(data)

    total = 0
    for x in data:
        total += (x - mean) ** 2

    return total / (len(data) - 1)


# Population Standard Deviation
def population_std(data):
    return population_variance(data) ** 0.5


# Sample Standard Deviation
def sample_std(data):
    return sample_variance(data) ** 0.5


# Four datasets
data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]

datasets = [data_1, data_2, data_3, data_4]

# Generate Report
for i, data in enumerate(datasets, start=1):

    print("\nDataset", i)
    print("Data:", data)

    print("Mean:", calculate_mean(data))
    print("Population Variance:", population_variance(data))
    print("Sample Variance:", sample_variance(data))
    print("Population SD:", population_std(data))
    print("Sample SD:", sample_std(data))
```

Q4. Investigating the Effect of Mathematical Transformations

```python
measurements = [12, 15, 18, 20, 25, 30, 40]
```

```python

measurements = [12, 15, 18, 20, 25, 30, 40]

# Mean
def calculate_mean(data):
    return sum(data) / len(data)

# Population Variance
def population_variance(data):
    mean = calculate_mean(data)

    return sum((x - mean) ** 2 for x in data) / len(data)

# Standard Deviation
def population_std(data):
    return population_variance(data) ** 0.5


# Transformation 1: Add 10
data_add = [x + 10 for x in measurements]

# Transformation 2: Multiply by 4
data_multiply = [x * 4 for x in measurements]


# Function to display statistics
def display_stats(name, data):

    print("\n", name)
    print("Mean:", calculate_mean(data))
    print("Variance:", population_variance(data))
    print("Standard Deviation:", population_std(data))


# Display all datasets
display_stats("Original Dataset", measurements)

display_stats("After Adding 10", data_add)

display_stats("After Multiplying by 4", data_multiply)


# Verify mathematical relationships
print("\nVerification")

print("Addition - Variance unchanged:",
      population_variance(data_add) == population_variance(measurements))

print("Multiplication - Mean:",
      calculate_mean(measurements) * 4)

print("Multiplication - Variance:",
      population_variance(measurements) * 16)

print("Multiplication - SD:",
      population_std(measurements) * 4)
```

Q5. Outlier Detection and Robust Spread Analysis

```python
sales = [
    12, 14, 15, 15, 16, 17, 18,
    19, 20, 21, 22, 24, 25, 120
]
```

```python

sales = [
    12, 14, 15, 15, 16, 17, 18,
    19, 20, 21, 22, 24, 25, 120
]

# Mean
def calculate_mean(data):
    return sum(data) / len(data)

# Median
def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 0:
        return (data[n//2-1] + data[n//2]) / 2

    return data[n//2]

# Population Variance
def population_variance(data):
    mean = calculate_mean(data)

    total = 0

    for x in data:
        total += (x - mean) ** 2

    return total / len(data)

# Standard Deviation
def population_std(data):
    return population_variance(data) ** 0.5


# Original statistics
print("Mean:", calculate_mean(sales))
print("Median:", calculate_median(sales))
print("Variance:", population_variance(sales))
print("Standard Deviation:", population_std(sales))


# IQR calculation
import pandas as pd

s = pd.Series(sales)

Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)


# Identify outliers
outliers = s[
    (s < lower_bound) |
    (s > upper_bound)
]

print("Outliers:", outliers.tolist())
```

(b) Exclude the flagged outlier and recalculate all measures.

```python

# Original statistics
old_mean = calculate_mean(sales)
old_median = calculate_median(sales)
old_variance = population_variance(sales)
old_std = population_std(sales)

# Remove the outlier
new_sales = [x for x in sales if x != 120]

# Recalculate statistics
new_mean = calculate_mean(new_sales)
new_median = calculate_median(new_sales)
new_variance = population_variance(new_sales)
new_std = population_std(new_sales)

print("After Removing Outlier")

print("Mean:", new_mean)
print("Median:", new_median)
print("Variance:", new_variance)
print("Standard Deviation:", new_std)


# Percentage changes
mean_change = ((new_mean-old_mean)/old_mean)*100

median_change = ((new_median-old_median)/old_median)*100

variance_change = ((new_variance-old_variance)/old_variance)*100

std_change = ((new_std-old_std)/old_std)*100


print("\nPercentage Changes")

print("Mean:", mean_change, "%")
print("Median:", median_change, "%")
print("Variance:", variance_change, "%")
print("Standard Deviation:", std_change, "%")
```
